A population is the complete group we want to study, while a sample is a smaller group selected from that population.

Task.2 :- Write a Python script that simulates random sampling: given a list of 1000 fake user IDs (e.g., user1 to user1000), select 50 random users to simulate a survey sample. Print the selected user IDs.<br><br><em><strong>Hint:</strong> Use the random.sample() function from Python's random module.</em>


```python
import random

users = [f"user{i}" for i in range(1, 1001)]

sample = random.sample(users, 50)

print("Selected survey users:")
for user in sample:
    print(user)
```

Studying every Swiggy order ever made would require an enormous amount of data and could be expensive and time-consuming to process. A properly selected sample of orders can provide a good estimate of overall delivery times while requiring much less time and computational effort.

Task.4 :- Given this scenario: A survey about music preferences is only sent to users who recently streamed Bollywood songs on Spotify. Identify and explain the type of bias this introduces in the sample.


The survey introduces selection bias (specifically, a form of sampling bias).

Users who recently streamed Bollywood songs are more likely to have Bollywood preferences, so they do not represent all Spotify users. The results would therefore likely overestimate the popularity of Bollywood music.

Task.5 :- Choose any two types of sampling (for example: random sampling, stratified sampling, systematic sampling, cluster sampling). For each, describe how you would use it to select a sample of users from BookMyShow to study movie booking habits.


A. Random Sampling

Create a list of BookMyShow users who made movie bookings and randomly select, for example, 500 users. Every user in the population has an equal chance of being selected, making the sample less likely to systematically favor a particular type of user.

B. Stratified Sampling

Divide users into groups (strata) based on characteristics such as age groups—e.g., 18–25, 26–40, and 41+. Then randomly select users from each group in proportion to its size. This ensures that different age groups are represented when studying movie-booking habits.

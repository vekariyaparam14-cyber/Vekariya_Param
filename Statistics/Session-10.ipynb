Task.1 :- Write a Python function bayes_posterior(prior, likelihood, evidence) that calculates and returns the posterior probability using Bayes theorem formula for given values.


```python
def bayes_posterior(prior, likelihood, evidence):
    posterior = (likelihood * prior) / evidence
    return posterior

result = bayes_posterior(0.30, 0.80,0.50)

print("Posterior probability:", result)
print("Percentage:", result * 100, "%")
```

Task.2 :- Given the following scenario: On Swiggy, 10% of restaurants are 'premium', and 80% of premium restaurants offer a discount, while only 20% of regular restaurants do. If you see a discount, use Bayes theorem in code to find the probability that the restaurant is premium.<br><br><em><strong>Hint:</strong> Define prior, likelihood, and evidence in your code with comments.</em>


```python
prior_premium = 0.10

likelihood_discount_premium = 0.80

likelihood_discount_regular = 0.20

prior_regular = 1 - prior_premium

evidence_discount = (
    likelihood_discount_premium * prior_premium
    + likelihood_discount_regular * prior_regular
)

posterior = (
    likelihood_discount_premium * prior_premium
) / evidence_discount

print("Probability restaurant is premium given discount:",posterior)

print("Percentage:", posterior * 100, "%")
```

Task.3 :- Simulate a basic spam detection filter: Given a list of 20 email subjects (some spam, some not), and the probability that 'free' appears in spam vs non-spam emails, write code to calculate the probability an email is spam if it contains 'free', using Bayes theorem.


```python
subjects = [
    "Win free money now",
    "Meeting at 10 AM",
    "Get free gift today",
    "Project report attached",
    "Free vacation offer",
    "Your electricity bill",
    "Congratulations, free prize",
    "Team meeting tomorrow",
    "Free shopping voucher",
    "Lunch plans",
    "Important bank update",
    "Free cash reward",
    "Assignment submission",
    "Weekend plans",
    "Free lottery ticket",
    "Office meeting",
    "Free mobile recharge",
    "Project discussion",
    "Free coupon",
    "Class schedule"
]

prior_spam = 8 / 20

likelihood_free_spam = 6 / 8

likelihood_free_not_spam = 2 / 12

prior_not_spam = 1 - prior_spam

evidence_free = (
    likelihood_free_spam * prior_spam
    + likelihood_free_not_spam * prior_not_spam
)

posterior_spam = (
    likelihood_free_spam * prior_spam
) / evidence_free

print("Probability email is spam if it contains 'free':",posterior_spam)

print("Percentage:", posterior_spam * 100, "%")
```

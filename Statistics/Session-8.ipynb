Task.2 :- Simulate rolling a six-sided dice 100 times using Python's random module, count how many times a '6' appears, and calculate the experimental probability of getting a '6'.<br><br><em><strong>Hint:</strong> Use random.randint(1,6) in a loop and keep a counter for '6'.</em>


```python
import random

count_six = 0

for i in range(100):
    roll = random.randint(1, 6)
    
    if roll == 6:
        count_six += 1

probability = count_six / 100

print("Number of times 6 appeared:", count_six)
print("Experimental probability of getting 6:", probability)
```

Task.4 :- Write a function that, given two arrays — one for users who ordered food on Zomato and one for users who ordered dessert — calculates the probability that a randomly selected user ordered both food and dessert (i.e., intersection over total unique users).


```python
def probability(food_user,desert_user):
    print("food user:",food_user)
    print("desert user:",desert_user)
    
    food_order = set(food_user)
    desert_order = set(desert_user)
    
    both = food_order.intersection(desert_order)
    total = food_order.union(desert_order)
    
    return print("probability of both is ", len(both)/len(total))

probability(["param","hardik","harsh","meet","harivadan","lakshya"],["param","hardik","mudra","minal"])
```

Task.5:- Explain with a short Python code example how conditional probability could be used to improve recommendations in an app like Flipkart (e.g., probability of buying headphones given user bought a phone).<br><br><em><strong>Hint:</strong> Use counts of users who bought both items vs total who bought a phone.</em>


```python
# Conditional probability helps an app like Flipkart predict what a user is 
# likely to buy next based on what they already bought.
# P(A|B) - it also probability of an event which is already occured

def conditional_probability(both_bought, phone_buyers):
    return both_bought / phone_buyers


probability = conditional_probability(30,100)

print("Probability of buying headphones given phone purchase:",probability)
print("Percentage:", probability * 100, "%")
```

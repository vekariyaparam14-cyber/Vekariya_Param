```python
import numpy as np
import matplotlib.pyplot as plt

mean = 50
std_dev = 10

data = np.random.normal(mean, std_dev, 100)

plt.hist(data, bins=10, density=True, alpha=0.7)

plt.xlabel("Values")
plt.ylabel("Density")
plt.title("Normal Distribution (Mean = 50, SD = 10)")

plt.show()
```

```python
import numpy as np
from scipy import stats

steps = [
    8000, 8200, 7900, 8100, 8000,
    8300, 8000, 8100, 8200, 8000,
    7900, 8100, 8000, 8300, 8200,
    8000, 8100, 7900, 8200, 8000
]

mean = np.mean(steps)

median = np.median(steps)

mode = stats.mode(steps, keepdims=True).mode[0]

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)


# doubt -->
if abs(mean - median) < 200 and abs(median - mode) < 200:
    print("Mean, median, and mode are approximately equal.")
    print("The data may be approximately normally distributed.")
else:
    print("Mean, median, and mode are not approximately equal.")
```

```python
import numpy as np

mean = 70
std_dev = 15

scores = np.random.normal(mean, std_dev, 200)

lower = mean - std_dev   # 55
upper = mean + std_dev   # 85

within_one_std = np.sum((scores >= lower) & (scores <= upper))

print("Total players:", len(scores))
print("Players scoring between 55 and 85:", within_one_std)
print("Percentage:", (within_one_std / 200) * 100, "%")
```

Task.2 :- Take the following list of Instagram influencer follower counts (in thousands): [5, 7, 8, 8, 9, 10, 12, 15, 95]. Calculate the mean and median, then state which is higher and what that tells you about the skewness of the data.<br><br><em><strong>Hint:</strong> If the mean is pulled away from the median, the data is skewed in that direction.</em>


```python
followers = [5, 7, 8, 8, 9, 10, 12, 15, 95]

mean = sum(followers) / len(followers)

followers.sort()

n = len(followers)
median = followers[n // 2]

print("Mean:", round(mean, 2), "thousand")
print("Median:", median, "thousand")

if mean > median:
    print("The data is right-skewed.")
elif mean < median:
    print("The data is left-skewed.")
else:
    print("The data is approximately symmetrical.")
```

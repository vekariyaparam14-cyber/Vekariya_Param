Task.3 :- Pick any one feature from an app you use (such as the number of likes on an Instagram post, or the rating of a restaurant on Zomato) and describe how statistics could help improve that feature for users.


Feature: Restaurant ratings on Zomato

Statistics could be used to calculate the average rating and the number of ratings for each restaurant. This would help users identify consistently well-rated restaurants while also preventing a restaurant with only a few ratings from appearing more reliable than one with thousands of ratings. The app could also analyze rating patterns to detect sudden changes in customer satisfaction.

Task.4 :- Suppose you are building a new playlist recommendation feature for a Spotify-like app. List 3 basic statistical terms (such as mean, median, mode, variance, etc.) that would be useful in analyzing user listening data, and explain each with a one-line example.


Mean (average): Find the average number of minutes a user listens to music each day—for example, 60 minutes/day.

Median: Find the middle value of users' daily listening times, which can be more representative when some users listen for extremely long periods.

Mode: Find the most frequently played genre—for example, if Pop is played most often, the app could recommend more Pop songs.

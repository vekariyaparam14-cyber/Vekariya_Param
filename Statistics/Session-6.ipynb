Task.1 :- Write down the null hypothesis (H0) and alternative hypothesis (H1) for the following scenario: You believe that people spend more time on Instagram than on YouTube daily.


Let:

\(\mu_I\) = average daily time spent on Instagram
\(\mu_Y\) = average daily time spent on YouTube

Null hypothesis (H₀):

$$ H_0:\mu_I \leq \mu_Y $$

There is no evidence that people spend more time on Instagram than YouTube.

Alternative hypothesis (H₁):

$$ H_1:\mu_I > \mu_Y $$

People spend more time on Instagram than YouTube.

Task.2 :- Given the data: Group A (users shown a new Zomato home page) had an average order value of ₹350 from 30 users, and Group B (old home page) had an average order value of ₹320 from 30 users. Assume both groups have a standard deviation of ₹50. Calculate the p-value using a two-sample t-test (you may use Python, R, or an online calculator), and decide whether to accept or reject the null hypothesis at a 0.05 significance level.<br><br><em><strong>Hint:</strong> Use scipy.stats.ttest_ind in Python or any online t-test calculator if you are not comfortable with code.</em>


```python
from scipy.stats import ttest_ind_from_stats

t_stat, p_value = ttest_ind_from_stats(
    mean1=350, std1=50, nobs1=30,
    mean2=320, std2=50, nobs2=30,
    equal_var=True
)

print("t-statistic:", t_stat)
print("p-value:", p_value)

"""
Decision

Since:

p=0.0236 < 0.05 

we reject H₀ at the 5% significance level.

Conclusion:
There is statistically significant evidence that the new Zomato home page is associated with a different average order value.
Since Group A's mean is higher, the result supports the idea that the new page increased average order value.
"""
```

Suppose we want to test whether a new dark-mode design improves user engagement.

Control group (A): Users see the existing dark mode.

Metric: Average session duration or percentage of users who return to the app the next day.

$$ H_1:\mu_B>\mu_A $$

Randomly assigning users to A and B helps ensure that differences in the metric can be attributed to the new design rather than systematic differences between the groups.

Task.4 :- Suppose you run an A/B test for a new 'Add to Wishlist' button design on Flipkart. Out of 1000 users, 520 clicked the new button versus 480 who clicked the old button. Without doing detailed calculations, explain if this difference is likely to be significant or just due to chance. Briefly explain how the p-value helps you decide.


Suppose:

New button: 520/1000 users clicked

Old button: 480/1000 users clicked

So the observed click rates are:

New: 52%

Old: 48%

Difference: 4 percentage points

A 4-point difference might be statistically significant, but we cannot determine that just by looking at the percentages. We need a statistical test to calculate the p-value.

The p-value tells us how likely it would be to observe a difference this large (or larger) if there were actually no real difference between the two buttons.

If p < 0.05 → reject H₀; evidence suggests the new button really performs differently.

If p ≥ 0.05 → fail to reject H₀; the observed 4% difference could reasonably be due to random chance.

So, the p-value—not the 52% vs 48% difference alone—helps us decide whether the result is statistically significant.

Task.1 :- Create a Python list to represent a playlist of 5 song durations in seconds (e.g., [210, 180, 245, 200, 175]). Print the total duration of the playlist.


```python
playlist = [210, 180, 245, 200, 175]

total_duration = sum(playlist)

print("Total playlist duration:", total_duration, "seconds")
```

Task.2 :- Represent a 3x3 matrix in Python to store the ratings (1-5) given by 3 friends to 3 different food items on Swiggy. Print the matrix and display the rating each friend gave to the second food item.


```python
ratings = [
    [5, 4, 3],  
    [4, 5, 4],  
    [3, 4, 5]   
]

print("Ratings matrix:")
for row in ratings:
    print(row)

print("\nRatings for the second food item:")

for i in range(len(ratings)):
    print("Friend", i + 1, ":", ratings[i][1])
```

Task.3 :- Given a matrix representing the number of orders for 4 different dishes across 3 days (rows: days, columns: dishes), write code to calculate and print the total orders for each dish.


```python
orders = [
    [10, 15, 20, 12],  
    [12, 18, 15, 10],  
    [8,  20, 17, 14]  
]

number_of_dishes = len(orders[0])

for dish in range(number_of_dishes):
    total = 0

    for day in range(len(orders)):
        total += orders[day][dish]

    print("Dish", dish + 1, "total orders:", total)
```

Task.4 :- Imagine you have a table of 5 Flipkart products with columns for price and quantity sold. Represent this data as a matrix in Python, then write a function to calculate the total revenue for each product.<br><br><em><strong>Hint:</strong> Multiply price and quantity for each row, then print the results.</em>


```python
products = [
    [500, 10],
    [1000, 8],
    [750, 15],
    [1200, 5],
    [300, 20]
]

def calculate_revenue(products):
    revenues = []

    for product in products:
        price = product[0]
        quantity = product[1]

        revenue = price * quantity
        revenues.append(revenue)

    return revenues


revenues = calculate_revenue(products)

for i, revenue in enumerate(revenues):
    print("Product", i + 1, "revenue: ₹", revenue)
```

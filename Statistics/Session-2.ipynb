Task.3 :- Write a function most_common_rating(ratings) that takes a list of Flipkart product ratings (e.g., [5, 4, 4, 3, 5, 4, 2, 4]) and returns the mode (most frequent rating).


```python
def most_common_rating(ratings):
    counts = Counter(ratings)
    return counts.most_common(1)[0][0]


ratings = [5, 4, 4, 3, 5, 4, 2, 4]
print("Most common rating:", most_common_rating(ratings))
```

Task.4 :- Given three lists representing YouTube video views for three different channels, compare the mean, median, and mode for each channel, and decide which channel's data is most affected by outliers.<br><br><em><strong>Hint:</strong> Try using one list with a very high value (e.g., [100, 120, 110, 105, 5000]) to see the effect on mean vs median.</em>


```python
channels = {
    "Channel A": [100, 120, 110, 105, 5000],  # Strong outlier
    "Channel B": [100, 120, 110, 105, 130],
    "Channel C": [80, 90, 100, 110, 120],
}

for name, views in channels.items():
    print(
        f"{name}: mean={mean(views):.2f}, "
        f"median={median(views)}, mode={mode(views) if len(set(views)) < len(views) else 'No mode'}"
    )

print("Channel A is most affected by outliers because 5000 is much larger")
print("than the other values, pulling its mean far above its median.")
```

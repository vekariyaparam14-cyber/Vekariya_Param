Task.1 :- Create two Python lists: one representing the number of hours you spend on Instagram each day for a week, and another for the number of posts you make each day. Plot these as a scatter plot using matplotlib and visually describe whether the relationship looks like a positive, negative, or zero correlation.


```python
import matplotlib.pyplot as plt

instagram_hours = [1, 2, 2.5, 3, 3.5, 4, 5]

posts = [1, 2, 2, 3, 4, 4, 6]

plt.scatter(instagram_hours, posts)

plt.xlabel("Hours spent on Instagram")
plt.ylabel("Number of posts")
plt.title("Instagram Usage vs. Posts")

plt.show() 

# Observation: The points generally move upward from left to right, so this shows a positive correlation.
#              As Instagram usage increases, the number of posts tends to increase.
```

Task.2 :- Given the following data for 6 friends: their monthly Zomato order count and their monthly Swiggy order count, calculate the correlation coefficient between the two lists using numpy's corrcoef(). Print the result and explain if the correlation is positive, negative, or zero.


```python
import numpy as np

zomato_orders = [5, 8, 10, 12, 15, 18]
swiggy_orders = [4, 7, 9, 13, 14, 17]

correlation_matrix = np.corrcoef(zomato_orders, swiggy_orders)

correlation = correlation_matrix[0, 1]

print("Correlation coefficient:", correlation)

if correlation > 0:
    print("The correlation is positive.")
elif correlation < 0:
    print("The correlation is negative.")
else:
    print("The correlation is zero.")

# Expected result: The coefficient will be close to +1, so the two variables have a strong positive correlation.
#                  Friends who order more frequently from Zomato also tend to order more frequently from Swiggy in this example.
```

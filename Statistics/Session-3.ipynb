Task.1 :- Given an array of daily step counts from your fitness app for one week (e.g., [4500, 7000, 5000, 8000, 4000, 9000, 6000]), calculate the variance by hand using the variance formula and show each calculation step.


```python
steps = [4500, 7000, 5000, 8000, 4000, 9000, 6000]

mean = sum(steps) / len(steps)

squared_deviations = [(x - mean) ** 2 for x in steps]

variance = sum(squared_deviations) / len(steps)

print("Mean:", mean)
print("Squared deviations:", squared_deviations)
print("Variance:", variance)
```

Task.2 :- Write a Python function called calculate_standard_deviation(scores) that takes a list of exam scores and returns the standard deviation rounded to two decimal places.


```python
import math

def calculate_standard_deviation(scores):
    mean = sum(scores) / len(scores)
    variance = sum((score - mean) ** 2 for score in scores) / len(scores)
    standard_deviation = math.sqrt(variance)
    return round(standard_deviation, 2)

scores = [70, 80, 90, 60, 100]
print(calculate_standard_deviation(scores))
```

Task.4 :- You are given the monthly salary data of 6 employees in a startup: [28000, 29000, 27000, 60000, 26500, 27500]. Calculate the variance and standard deviation, then interpret in 2-3 lines what the high or low spread means for this team.


```python
import math

salaries = [28000, 29000, 27000, 60000, 26500, 27500]

mean = sum(salaries) / len(salaries)

squared_deviations = [(x - mean) ** 2 for x in salaries]

variance = sum(squared_deviations) / len(salaries)

standard_deviation = math.sqrt(variance)

print("Mean salary:", mean)
print("Variance:", variance)
print("Standard deviation:", standard_deviation)
```

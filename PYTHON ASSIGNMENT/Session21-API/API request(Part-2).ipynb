Task.1 :- Use the requests library to send a POST request to https://jsonplaceholder.typicode.com/posts with a JSON payload containing a title, body, and userId, then print the response status code and JSON data.


```python
import requests
url = "https://jsonplaceholder.typicode.com/posts"

data = {
    "title": "My First Post",
    "body" : "Hello from Python!",
    "userId" : 101
}
response = requests.post(url, json = data)
print(response)
print(response.status_code)
print(response.json())
```

Task.2 :- Build a Python script that lets a user enter a new playlist name and description, sends this data as JSON in a POST request to a mock API endpoint (such as https://jsonplaceholder.typicode.com/posts), and prints the playlist ID returned by the API.


```python
import requests

playlist_name = input("Enter playlist name: ")
playlist_description = input("Enter playlist description: ")

url = "https://jsonplaceholder.typicode.com/posts"

playlist_data = {
    "name": playlist_name,
    "description": playlist_description
}

response = requests.post(url, json=playlist_data)
data = response.json()

print("Playlist ID:", data["id"])
```

Task.3 :- Send a POST request to https://reqres.in/api/users with a JSON object containing a username and job, then parse the response to extract and print the created user's ID and creation timestamp.<br><br><em><strong>Hint:</strong> Use response.json() to access the returned data.</em>


```python
import requests

url = "https://reqres.in/api/users"

user_data = {
    "username": "Param",
    "job": "Data Scientist"
}

response = requests.post(url, json=user_data)
data = response.json()

print("User ID:", data["id"])
print("Created At:", data["createdAt"])
```

Task.4 :- Write a script that fetches the latest 5 posts from https://jsonplaceholder.typicode.com/posts, parses the JSON response, and saves the post titles and userIds to a CSV file called posts.csv.<br><br><em><strong>Hint:</strong> Use the csv module for writing to CSV.</em>


```python
# doubt
import requests
import csv

url = "https://jsonplaceholder.typicode.com/posts"

response = requests.get(url)

if response.status_code == 200:
    posts = response.json()

    latest_posts = posts[-5:]

    with open("posts.csv", "w", newline="", encoding="utf-8") as file:
        writer = csv.writer(file)

        writer.writerow(["userId", "title"])

        for post in latest_posts:
            writer.writerow([post["userId"], post["title"]])

    print("Latest 5 posts saved successfully to posts.csv")

else:
    print("Failed to fetch posts. Status code:", response.status_code)
```

Task.5 :- Modify your script to save the same API data (latest 5 posts from https://jsonplaceholder.typicode.com/posts) into a JSON file named posts.json instead of CSV.


```python
# doubt
import requests
import json

url = "https://jsonplaceholder.typicode.com/posts"

response = requests.get(url)

if response.status_code == 200:
    posts = response.json()

    latest_posts = posts[-5:]

    with open("posts.json", "w", encoding="utf-8") as file:
        json.dump(latest_posts, file, indent=4)

    print("Latest 5 posts saved successfully to posts.json")

else:
    print("Failed to fetch posts. Status code:", response.status_code)
```

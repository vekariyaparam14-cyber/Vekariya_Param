Task.1 :- Use the requests.Session() object to fetch your Flipkart order history page twice in a row (without logging in), and print the response status codes for both requests.<br><br><em><strong>Hint:</strong> Observe if cookies or session headers change between requests.</em>


```python
import requests

session = requests.Session()

url = "https://www.flipkart.com/account/orders"

response1 = session.get(url)

print("First Request Status Code:", response1.status_code)
print("Cookies after first request:", session.cookies.get_dict())

response2 = session.get(url)

print("Second Request Status Code:", response2.status_code)
print("Cookies after second request:", session.cookies.get_dict())
```

Task.2 :- Write a Python script using requests to call the OpenWeatherMap API (https://api.openweathermap.org/data/2.5/weather) for the city 'Ahmedabad' using your own API key, and print the current temperature.


```python
# doubt
import requests

API_KEY = "vekariyaparam"

url = "https://api.openweathermap.org/data/2.5/weather"

params = {
    "q": "Ahmedabad",
    "appid": API_KEY,
    "units": "metric"  # Temperature in Celsius
}

response = requests.get(url, params=params)

if response.status_code == 200:
    data = response.json()

    temperature = data["main"]["temp"]

    print("Current temperature in Ahmedabad:", temperature, "°C")

else:
    print("Error:", response.status_code)
    print(response.json())
```

Task.3 :- Simulate an async data fetch from two different APIs (for example, fetch trending songs from Spotify and trending movies from BookMyShow) using Python's asyncio and httpx library, and print both results when done.<br><br><em><strong>Hint:</strong> Use asyncio.gather() to run both requests concurrently.</em>

```python
# doubt
import asyncio
import httpx


async def fetch_data(client, url, name):
    try:
        response = await client.get(url)
        response.raise_for_status()

        data = response.json()
        print(f"\n{name} Data:")
        print(data[:3] if isinstance(data, list) else data)

    except httpx.HTTPError as e:
        print(f"Error fetching {name}: {e}")


async def main():
    # Example public APIs
    songs_url = "https://jsonplaceholder.typicode.com/posts"
    movies_url = "https://jsonplaceholder.typicode.com/users"

    async with httpx.AsyncClient() as client:
        # Run both API requests concurrently
        await asyncio.gather(
            fetch_data(client, songs_url, "Trending Songs"),
            fetch_data(client, movies_url, "Trending Movies")
        )


# Run the async program
asyncio.run(main())
```

Task.4 :- Many APIs require Bearer tokens for authentication. Write a function get_user_profile() that calls a mock API endpoint (e.g., https://jsonplaceholder.typicode.com/users/1) using a fake Bearer token in the Authorization header, and prints the user's name.<br><br><em><strong>Constraint:</strong> Use the 'Authorization: Bearer <token>' header format.</em>


```python
import requests


def get_user_profile():
    url = "https://jsonplaceholder.typicode.com/users/1"

    token = "fake_bearer_token_12345"

    headers = {
        "Authorization": f"Bearer {token}"
    }

    try:
        response = requests.get(url, headers=headers)
        response.raise_for_status()

        user = response.json()

        print("User Name:", user["name"])

    except requests.exceptions.RequestException as e:
        print("Error:", e)


get_user_profile()
```

Task.5 :- Use ChatGPT or Copilot to generate Python code that demonstrates the first step of an OAuth 2.0 login flow (for example, generating the URL to redirect a user to Spotify's OAuth login page). Paste the generated code and briefly explain what it does.


```python
import urllib.parse

# Spotify OAuth details
CLIENT_ID = "YOUR_SPOTIFY_CLIENT_ID"
REDIRECT_URI = "http://localhost:8000/callback"

# OAuth scopes (permissions requested)
SCOPES = [
    "user-read-email",
    "user-read-private"
]

# Spotify authorization endpoint
AUTH_URL = "https://accounts.spotify.com/authorize"

# Parameters for OAuth login
params = {
    "client_id": CLIENT_ID,
    "response_type": "code",
    "redirect_uri": REDIRECT_URI,
    "scope": " ".join(SCOPES)
}

# Generate the complete authorization URL
authorization_url = f"{AUTH_URL}?{urllib.parse.urlencode(params)}"

print("Redirect the user to this URL:")
print(authorization_url)
```

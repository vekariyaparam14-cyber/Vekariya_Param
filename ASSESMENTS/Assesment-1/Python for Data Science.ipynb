Question 2 - The dataset has columns like age, cholesterol, and sex. What is the difference between a numerical feature and a categorical feature?

```python
# Difference Between Numerical and Categorical Features

# Numerical Feature:
# - Contains numbers.
# - Can be measured or calculated.
# - Example: Age, Salary, Marks.

# Categorical Feature:
# - Contains categories or labels.
# - Cannot be measured mathematically.
# - Example: Gender, City, Color.

# Example:
# Age = 20          -> Numerical Feature
# Gender = Male     -> Categorical Feature
```

Question 3 - You run type() on the target column and it returns int. Why does the data type of your target variable matter before framing a prediction problem?

```python
# The data type of the target variable matters because it determines
# the type of prediction problem.

# If target is int/float → Regression (predict a number)
# Example: Salary = 50000

# If target is category/label → Classification
# Example: Pass/Fail

# Therefore, checking the target data type helps choose the correct
# machine learning model.
```

Task.2 :- Excecute a comprehensive outlier detection strategy on 'chol' (cholesterol) and
'thalach' (max heart rate) using the Interquartile Range (IQR) method and
visualize the results via Boxplots.

```python
df = pd.read_csv("heart.csv")

columns = ["chol","thalach"]

for col in columns:
    print(f"\n{col}:")
    
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_limit = Q1 - 1.5 * IQR
    higher_limit = Q3 + 1.5 * IQR
    
    outlier = df[(df[col] < lower_limit)| (df[col]> higher_limit)]
    print(f"Q1:",Q1)
    print(f"Q2:",Q3)
    print(f"IQR:",IQR)
    print(f"LOWER LIMIT:",lower_limit)
    print(f"HIGHER LIMIT:",higher_limit)
    print(f"outlier", len(outlier))


plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.boxplot(df["chol"])
plt.title("Cholestrol")

plt.subplot(1,2,2)
plt.boxplot(df["thalach"])
plt.title("Max. Heart Rate")
```

Task.4 :- Develop a feature engineering script to normalize numerical data using
StandardScaler and create derived features such as 'Age-adjusted Max Heart

```python
# Import libraries
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Load the dataset
df = pd.read_csv("heart.csv")   # Replace with your dataset path

# -----------------------------
# Step 1: Normalize Numerical Features
# -----------------------------

# Numerical columns
numerical_cols = [
    'age', 'trestbps', 'chol',
    'thalach', 'oldpeak'
]

# Initialize StandardScaler
scaler = StandardScaler()

# Scale the numerical columns
df_scaled = df.copy()
df_scaled[numerical_cols] = scaler.fit_transform(df[numerical_cols])

print("Normalized Numerical Features:")
print(df_scaled[numerical_cols].head())

# -----------------------------
# Step 2: Feature Engineering
# -----------------------------

# Maximum Predicted Heart Rate
df_scaled['predicted_max_hr'] = 220 - df['age']

# Age-adjusted Maximum Heart Rate
df_scaled['age_adjusted_max_hr'] = (
    df['thalach'] / (220 - df['age'])
)

# Cholesterol to Age Ratio
df_scaled['chol_age_ratio'] = (
    df['chol'] / df['age']
)

# Blood Pressure to Age Ratio
df_scaled['bp_age_ratio'] = (
    df['trestbps'] / df['age']
)

print("\nDerived Features:")
print(df_scaled[['predicted_max_hr',
                 'age_adjusted_max_hr',
                 'chol_age_ratio',
                 'bp_age_ratio']].head())

# -----------------------------
# Save Engineered Dataset
# -----------------------------
df_scaled.to_csv("heart_feature_engineered.csv", index=False)

print("\nFeature engineering completed successfully.")
print("Dataset saved as 'heart_feature_engineered.csv'")
```

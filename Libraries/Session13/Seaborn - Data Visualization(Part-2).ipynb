Task.2 :- Using the 'flights' dataset from Seaborn, generate a heatmap showing the correlation between months and years based on the number of passengers.<br><br><em><strong>Hint:</strong> Use pivot_table to reshape the data before plotting the heatmap.</em>


```python
flight = sns.load_dataset("flights")
print(flight)
flights_pivot = flight.pivot_table(
    values="passengers",
    index="month",
    columns="year"
)
```

```python
plt.figure(figsize=(15,6))
sns.heatmap(flights_pivot, annot=True, cmap="coolwarm")
```

Task.3 :-Create a relplot using the 'fmri' dataset from Seaborn to visualize how the signal changes over time for different event types.


```python
fri = sns.load_dataset("fmri")
fri
```

```python
plt.figure(figsize=(20,6))
sns.relplot(data=fri,x="timepoint",y="signal",kind="line",hue="event")
plt.show()
```

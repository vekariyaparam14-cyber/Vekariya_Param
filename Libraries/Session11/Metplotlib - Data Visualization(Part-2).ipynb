Task.2 :- Plot a comparison of average delivery times for Zomato, Swiggy, and Domino's using a bar chart, and style each bar with a different color, linestyle, and linewidth using Matplotlib plot styling options.

Task.3 :- Create a multi-axis chart that shows the number of Instagram followers (left y-axis) and average daily posts (right y-axis) for five influencers. Use different colors and markers for each axis.


```python
a = ["Param","Hardik","Harivadan","Meet","Parth"]
y1 = [350,400,300,450,500]
y2 = [1,5,3,4,2]

fig, ax = plt.subplots()

ax.plot(a,y1, label = "Followers", color = "red")

ax1 = ax.twinx()

ax1.plot(a,y2,label = "Daily Post", color ="green")

lines1, label1 = ax.get_legend_handles_labels()
lines2, label2 = ax1.get_legend_handles_labels()

ax.legend(lines1 + lines2, label1 + label2, loc = "upper center")
plt.show()
```

Task.4 :- Plot the number of tickets sold for five recent Bollywood movies (categorical) versus their IMDB ratings (numeric) using a scatter plot. Add annotations to display the movie names above each point.


```python
a = ["Pushpa","KGF","Dhurandhar","Salar","Marco"]
y1 = [25000,50000,15000,20000,10000]
y2 = [4.8,4.8,4.9,4.6,4.5]

fig, ax = plt.subplots()

ax.scatter(a,y1, label = "NO. of Ticket", color = "red")
ax.annotate("Pushpa", xy = ["Pushpa",25000])
ax.annotate("KGF", xy = ["KGF",50000])
ax.annotate("Dhurandhar", xy = ["Dhurandhar",15000])
ax.annotate("Salar", xy = ["Salar",20000])
ax.annotate("Marco", xy = ["Marco",10000])
ax1 = ax.twinx()

ax1.scatter(a,y2,label = "IMBD rate", color ="green")
ax1.annotate("Pushpa", xy = ["Pushpa",4.8])
ax1.annotate("KGF", xy = ["KGF",4.8])
ax1.annotate("Dhurandhar", xy = ["Dhurandhar",4.9])
ax1.annotate("Salar", xy = ["Salar",4.6])
ax1.annotate("Marco", xy = ["Marco",4.5])

lines1, label1 = ax.get_legend_handles_labels()
lines2, label2 = ax1.get_legend_handles_labels()

ax.legend(lines1 + lines2, label1 + label2, loc = "upper right")
plt.show()
```

Task.5 :- Add a custom text annotation to a Matplotlib chart showing Flipkart's monthly sales, marking the highest sales point with the label 'Big Billion Days'.<br><br><em><strong>Hint:</strong> Use the ax.annotate() function to place the label at the correct data point.</em>

```python
x = ["January","February","March","April","May","June","July","August","September","October","November","December"]
y = [15000,17000,8500,9500,10000,15000,20000,21000,50000,35000,45000,6000]

ax = plt.plot(x,y,marker = "o",color = "red")
plt.title("Flipcart Monthly Sales")
plt.ylabel("NO. of Sales")
plt.xlabel("Months")
max_sales = max(y)
index_max = y.index(max_sales)
plt.annotate(text =  "Big Billion Days", xy = [x[index_max],max_sales], xytext= (8.5,54000),color = "green",arrowprops=dict(facecolor='blue', arrowstyle='->'))
plt.show()
```

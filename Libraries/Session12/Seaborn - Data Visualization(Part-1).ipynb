Task.5 :- Use Seaborn's kdeplot to visualize the distribution of daily step counts for a week, as if analyzing data from a fitness app. Apply the 'whitegrid' theme and customize the plot color.


```python
a = pd.DataFrame({
    "Days": ["Monday","Tuesday","Wednesday","Thrusday","Friday","Saturday","Sunday"],
    "Steps":[10000,9500,8000,9900,6000,7500,6500]
    })
```

```python
sns.set_style("whitegrid")
plt.figure(figsize=(8,6))
sns.kdeplot(a["Steps"], color = "green", fill = True)
```
